```python
import pandas as pd

# Sample data
data = {
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Sales": [10000, 12000, 15000, 13000, 17000, 16000],
    "Profit": [2000, 3000, 4000, 2500, 3500, 3000]
}

df = pd.DataFrame(data)
```

```python
# 1- line plot sales over time

import matplotlib.pyplot as plt
%matplotlib inline

# Line Plot: Sales over Time
plt.figure(figsize=(8, 5))
plt.plot(df['Month'], df['Sales'], color='blue', marker='o', linestyle='-', label='Sales')
plt.title('Sales Trend Over Months')
plt.xlabel('Month')
plt.ylabel('Sales ($)')
plt.grid('True')
plt.legend()
plt.tight_layout()
plt.show()

# Output: A line plot showing the sales trend over the months.
```

```python
# Bar Plot: Sales vs Profit by Month
plt.figure(figsize=(10, 6))
width = 0.3
plt.bar(df['Month'], df['Sales'], width=width, label='Sales', color='skyblue')
plt.bar(df['Month'], df['Profit'], width=width, label='Profit', color='orange', bottom=df['Sales'])
plt.title('Sales and Profit Comparison by Month')
plt.xlabel('Month')
plt.ylabel('Amount ($)')
plt.legend()
plt.tight_layout()
plt.show()

# A stacked bar plot where you can compare Sales and Profit for each month.
#stacked bar
```

```python
plt.figure(figsize=(10, 6))
width = 0.3
plt.bar(df['Month'], df['Sales'], width=width, label='Sales', color='skyblue')
plt.bar(df['Month'], df['Profit'], width=width, label='Profit', color='orange', bottom=df['Sales'])
# in comparision of two graphs, always mention bottom='' to stack the bars on top of each other
# in place of x and y values, we use df['Month'] instead Month and df['Sales'] instead of Sales, and df['Profit'] instead of Profit,
  # because vaalues are stored in the dataframe df, and we need to access them using the column names.

plt.title('Sales and Profit Comparison by Month')
plt.xlabel('Month')
plt.ylabel('Amount ($)')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Pie Chart: Profit Distribution by Month
plt.figure(figsize=(5, 5))
plt.pie(df['Profit'], labels=df['Month'], autopct='%1.1f%%', startangle=100, colors=plt.cm.Paired.colors)
plt.title('Profit Distribution by Month')
plt.tight_layout()
plt.axis('equal')
plt.show()

# A pie chart displaying the proportion of profit distribution for each month.
```

```python
#pie

plt.pie(df['Profit'], labels=df['Month'] , autopct='%1.1f%%', startangle=120)
plt.title('Profit distribution by month')
plt.axis('equal')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Scatter Plot: Sales vs Profit (Correlation)
plt.figure(figsize=(8, 5))
plt.scatter(df['Sales'], df['Profit'], color='green', s=100, edgecolors='black')
# here s=100 is the size of the points, and edgecolors='black' adds a black border around the points for better visibility.
plt.title('Sales vs Profit')
plt.xlabel('Sales ($)')
plt.ylabel('Profit ($)')
plt.grid(True)
plt.tight_layout()
plt.show()

# A scatter plot showing the correlation between Sales and Profit.
```

```python
# Histogram: Distribution of Sales
plt.figure(figsize=(8, 5))
plt.hist(df['Sales'], bins=5, color='purple', edgecolor='black')
plt.title('Sales Distribution')
plt.xlabel('Sales ($)')
plt.ylabel('Frequency')
plt.tight_layout()
plt.show()

# A histogram showing the distribution of Sales values across months.
```

```python
# Box Plot: Profit Distribution
plt.figure(figsize=(8, 5))
plt.boxplot(df['Profit'], vert=True, patch_artist=True, boxprops=dict(facecolor="lightgreen"))
# here vert=True means the box plot is vertical, patch_artist=True allows us to fill the box with color, and boxprops=dict(facecolor="lightgreen") sets the color of the box to light green.
plt.title('Profit Distribution')
plt.xlabel('Profit ($)')
plt.tight_layout()
plt.show()

# A box plot showing the spread and outliers of Profit.
# in the below box plt, we can see that the median profit is around $3000, 
# and there are no significant outliers in the profit data. 
# The interquartile range (IQR) indicates that most of the profit values are concentrated between $2500 and $3500.
```
