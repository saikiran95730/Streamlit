```python
import numpy as np
import matplotlib.pyplot as plt 
%matplotlib inline
```

```python
# line plot
x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.figure(figsize=(8, 4))
plt.plot(x, y, label="sin(x)", color='blue')
plt.title("Sine Function")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.grid(False)
plt.legend()
plt.show()
```

```python
# bar chart

cities = ['Delhi', 'Mumbai','Bangalore', 'Chennai']
population = [18, 20, 12, 8]
plt.figure(figsize=(6,4))
plt.bar(cities, population, color='green')
plt.title("Population of Cities (in millions)") 
plt.xlabel("Cities")
plt.ylabel("Population")
plt.show()
```

```python
# pie chart

labels = ['Sleep','work', 'exercise','gym', 'movies']
sizes = [9,7,4,2,8]
colors = ['#ff9999','#66b3ff','#99ff99','#ffcc99','#c2c2f0']

plt.figure(figsize=(6,6))
plt.pie(sizes, labels=labels, colors=colors, autopct='%1.1f%%', startangle=200)
plt.title('Daily time spent by me')
plt.axis('equal')
plt.show()
```

```python
#histogram

scores = np.random.normal(loc=65, scale=10, size=1000) #size = no. of random values generated, loc = mean, scale = std deviation

plt.figure(figsize=(8,4))
plt.hist(scores, bins=10, color='green', edgecolor='red')
plt.title('Student scores')
plt.xlabel('Scores')
plt.ylabel('no. of stds')
plt.grid('True')
plt.show()
```

```python
#heatmap
data = np.random.rand(5, 5)

plt.figure(figsize=(6, 5))
plt.imshow(data, cmap='hot', interpolation='nearest')
plt.colorbar(label='Value')
plt.title("Random Heatmap")
plt.show()
```
